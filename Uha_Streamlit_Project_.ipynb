```python
%%writefile app.py

import streamlit as st

# Page configuration
st.set_page_config(
    page_title="Student Marks Calculator",
    page_icon="📚",
    layout="centered"
)

# Title
st.title("📚 Student Marks Calculator")
st.write("Enter your marks below to calculate your total, average and result.")

# Student name
name = st.text_input("Enter Student Name")

# Marks input
st.subheader("Enter Subject Marks")

subject1 = st.number_input("Subject 1", min_value=0, max_value=100, value=0)
subject2 = st.number_input("Subject 2", min_value=0, max_value=100, value=0)
subject3 = st.number_input("Subject 3", min_value=0, max_value=100, value=0)
subject4 = st.number_input("Subject 4", min_value=0, max_value=100, value=0)
subject5 = st.number_input("Subject 5", min_value=0, max_value=100, value=0)

# Calculate button
if st.button("Calculate Results"):

    if name.strip() == "":
        st.error("Please enter the student name.")

    else:
        marks = [
            subject1,
            subject2,
            subject3,
            subject4,
            subject5
        ]

        total = sum(marks)
        average = total / 5
        percentage = (total / 500) * 100

        # Display results
        st.subheader("📊 Student Results")

        st.write("Student Name:", name)
        st.write("Total Marks:", total, "/ 500")
        st.write("Average Marks:", round(average, 2))
        st.write("Percentage:", round(percentage, 2), "%")

        # Pass or fail
        if all(mark >= 35 for mark in marks):
            st.success("Result: PASS 🎉")
        else:
            st.error("Result: FAIL ❌")

        # Display individual marks
        st.subheader("Subject-wise Marks")

        for i, mark in enumerate(marks, start=1):
            st.write(f"Subject {i}: {mark}/100")
```
